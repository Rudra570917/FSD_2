```python
import plotly.express as px
import pandas as pd
import numpy as np
import seaborn as sns
import plotly.io as pio
pio.renderers.default="notebook"
```

```python
df=pd.read_csv("social_media_engagement.csv")
```

```python
df
```

```python
df.columns
```

```python
fig=px.bar(data_frame=df,x="post_type",y="likes",text_auto=True,color="comments",color_discrete_map={"Facebook":"yellow","Instagram":"red","Twitter":"black"},height=700,width=2000,title="Platform",hover_data=["post_day","sentiment_score"],facet_col="sentiment_score",facet_row="post_day")
fig.show()
```

```python
px.bar(data_frame=df,x="post_type",y=["likes","comments","shares"],color="platform",barmode="group",pattern_shape="sentiment_score",text="comments")
```
